```python
# Revisar valores atípicos en la duración

plt.boxplot(df["duration_ms"])

plt.title("Valores atípicos en la duración de las canciones")
plt.ylabel("Duración en milisegundos")

plt.show()
```

```python
# Calcular los cuartiles
Q1 = df["duration_ms"].quantile(0.25)
Q3 = df["duration_ms"].quantile(0.75)

# Calcular el rango intercuartílico
IQR = Q3 - Q1

# Encontrar los valores atípicos
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

outliers = df[(df["duration_ms"] < limite_inferior) |
              (df["duration_ms"] > limite_superior)]

print("Cantidad de valores atípicos:", len(outliers))
print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)
```

```python
# Revisar valores atípicos en la popularidad

plt.boxplot(df["popularity"])

plt.title("Valores atípicos en la popularidad")
plt.ylabel("Popularidad")

plt.show()
```

```python
# Calcular los cuartiles
Q1 = df["popularity"].quantile(0.25)
Q3 = df["popularity"].quantile(0.75)

# Calcular el rango intercuartílico
IQR = Q3 - Q1

# Calcular los límites
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Encontrar los valores atípicos
outliers = df[(df["popularity"] < limite_inferior) |
              (df["popularity"] > limite_superior)]

print("Cantidad de valores atípicos:", len(outliers))
print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)
```

```python
# Comparar los valores atípicos encontrados

print("Outliers en duración:", 107)
print("Outliers en popularidad:", 36)
```

```python
# Crear una columna con la duración en minutos

df["duration_min"] = df["duration_ms"] / 60000

print(df[["track_name", "duration_ms", "duration_min"]].head())
```

```python
# Crear una columna para identificar canciones populares
promedio = df["popularity"].mean()

df["es_popular"] = df["popularity"] >= promedio

print(df[["track_name", "popularity", "es_popular"]].head())
```

```python
# Agrupar las canciones por género y calcular su popularidad promedio
popularidad_genero = df.groupby("genre")["popularity"].mean()

print(popularidad_genero)
```

```python
# Contar la cantidad de canciones por género
cantidad_genero = df.groupby("genre")["track_name"].count()

print(cantidad_genero)
```
